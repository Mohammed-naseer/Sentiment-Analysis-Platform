# 07. Ensemble & Decision Layer Evaluation

### Approaches Tested:
1. **Majority Voting**: Unweighted discrete voting across CNN, LSTM, RoBERTa.
2. **Confidence-Weighted Voting**: Soft ensemble weighting model probabilities by confidence scores.
3. **Disagreement Analysis**: Diagnostic inspection on conflicting test cases.

```python
import os
import sys

sys.path.append(os.path.abspath('..'))
from src.ensemble.decision_layer import DecisionLayer

ensemble_weighted = DecisionLayer(strategy="confidence_weighted")
ensemble_majority = DecisionLayer(strategy="majority_voting")

# Sample Case: Conflicting predictions
conflicting_preds = {
    "cnn": {"label": "Mixed", "confidence": 0.74, "probabilities": {"Positive": 0.15, "Negative": 0.11, "Mixed": 0.74}},
    "lstm": {"label": "Negative", "confidence": 0.61, "probabilities": {"Positive": 0.10, "Negative": 0.61, "Mixed": 0.29}},
    "roberta": {"label": "Mixed", "confidence": 0.88, "probabilities": {"Positive": 0.05, "Negative": 0.07, "Mixed": 0.88}}
}

res_weighted = ensemble_weighted.combine(conflicting_preds)
res_majority = ensemble_majority.combine(conflicting_preds)

print("=== Confidence-Weighted Ensemble Result ===")
print("Sentiment:", res_weighted["sentiment"])
print("Confidence:", res_weighted["confidence"])
print("Agreement %:", res_weighted["agreement_percentage"])
print("Disagreement Reason:", res_weighted["disagreement_reason"])

print("\n=== Majority Voting Ensemble Result ===")
print("Sentiment:", res_majority["sentiment"])
print("Confidence:", res_majority["confidence"])
print("Agreement %:", res_majority["agreement_percentage"])
```
