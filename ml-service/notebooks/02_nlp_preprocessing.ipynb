# 02. NLP Preprocessing Pipeline

### Goal:
- Normalize Unicode and expand English contractions.
- Clean noise, URLs, and HTML without removing crucial sentiment negations (`not`, `never`, `no`).
- Construct vocabulary index for CNN/LSTM and HuggingFace tokenization for RoBERTa.

```python
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath('..'))
from src.preprocessing.nlp_pipeline import clean_text, Vocabulary, LabelEncoder, NEGATION_WORDS

sample_sentences = [
    "I can't believe this is NOT working!",
    "Super fast delivery & great value for money https://example.com",
    "Never buying from this seller again... 100% disappointing!"
]

print("Preserved Negation Words:", list(NEGATION_WORDS)[:10])
print("\n=== Before vs After Cleaning ===")
for s in sample_sentences:
    print(f"Original: {s}")
    print(f"Cleaned : {clean_text(s)}\n")
```

```python
# Build Vocabulary
df = pd.read_csv('../data/sample_sentiment_benchmark.csv')
cleaned_texts = df['text'].apply(clean_text).tolist()

vocab = Vocabulary(min_freq=1, max_size=10000)
vocab.build_vocab(cleaned_texts)
print(f"Vocabulary size: {len(vocab)} unique tokens")

# Sequence transformation
demo_seq = vocab.text_to_sequence("great sound quality", max_len=10)
print("Integer Sequence for 'great sound quality':", demo_seq)
```

```python
# Label Encoder
le = LabelEncoder()
encoded_labels = le.fit_transform(df['sentiment'].tolist())
print("Classes:", le.classes_)
print("Encoded:", encoded_labels[:5])
print("Decoded:", le.inverse_transform(encoded_labels[:5]))
```
