# 01. Dataset Exploration & Validation

### Goal:
- Load CSV dataset and automatically detect text/label/emotion columns.
- Profile dataset statistics: total rows, missing values, duplicate records, text length distributions, class imbalance.
- Validate data integrity and split cleanly into train/validation/test sets (preventing data leakage).

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add ml-service root to Python path
sys.path.append(os.path.abspath('..'))
from src.preprocessing.nlp_pipeline import DatasetValidator

DATASET_PATH = '../data/sample_sentiment_benchmark.csv'
df = pd.read_csv(DATASET_PATH)
print(f"Loaded {len(df)} rows from {DATASET_PATH}")
df.head()
```

```python
# Profile Dataset Integrity
profile = DatasetValidator.validate_and_profile(df)
print("=== Dataset Profile Report ===")
for k, v in profile.items():
    print(f"{k}: {v}")
```

```python
# Visualize Class & Emotion Distribution
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
df['sentiment'].value_counts().plot(kind='bar', color='skyblue', title='Sentiment Class Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')

if 'emotion' in df.columns:
    plt.subplot(1, 2, 2)
    df['emotion'].value_counts().plot(kind='bar', color='salmon', title='Emotion Distribution')
    plt.xlabel('Emotion')
    plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Stratified Data Splitting
train_df, val_df, test_df = DatasetValidator.prepare_splits(
    df, text_col='text', label_col='sentiment', test_size=0.2, val_size=0.2, random_state=42
)
print(f"Training set: {len(train_df)} rows")
print(f"Validation set: {len(val_df)} rows")
print(f"Test set: {len(test_df)} rows")
```
